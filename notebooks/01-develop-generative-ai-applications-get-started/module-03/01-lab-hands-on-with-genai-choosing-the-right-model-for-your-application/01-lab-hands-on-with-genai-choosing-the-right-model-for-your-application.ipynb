```python
from os import getenv
from dotenv import load_dotenv
from json import dumps

from langchain_openrouter import ChatOpenRouter 

load_dotenv()

OPENROUTER_API_KEY = getenv("OPENROUTER_API_KEY")
OPENROUTER_DEFAULT_BASE_URL = "https://openrouter.ai/api/v1"
OPENROUTER_MODEL = "ibm-granite/granite-4.1-8b"

def llm_model(params=None):

    # 1. Define sensible defaults
    config = {
        "model": OPENROUTER_MODEL,
        "api_key": OPENROUTER_API_KEY,
        "base_url": OPENROUTER_DEFAULT_BASE_URL,
        "temperature": 0.5,
        "max_tokens": 256,
        "max_completion_tokens": 128
    }
    
    if params:
        config.update(params)
        
    # 3. Initialize the model
    model = ChatOpenRouter(
        model=config["model"],
        api_key=config["api_key"],
        base_url=config["base_url"],
        temperature=config["temperature"],
        max_tokens=config["max_tokens"],
        max_completion_tokens=config["max_completion_tokens"]
    )

    return model

def llm_model_response(prompt_text, params=None):

    model = llm_model(params)
    response = model.invoke(prompt_text)
    
    return response
```

```python
params = {
    "model": "openai/gpt-4o-mini",
    "api_key": OPENROUTER_API_KEY,
    "base_url": OPENROUTER_DEFAULT_BASE_URL,
    "temperature": 0.8,
    "max_tokens": 1024,
    "max_completion_tokens": 512
}

text = "Only reply with the answer. What is the capital of Canada?"

response_gpt_4o_mini = llm_model_response(text, params)
params = {
    "model": "meta-llama/llama-4-maverick",
    "api_key": OPENROUTER_API_KEY,
    "base_url": OPENROUTER_DEFAULT_BASE_URL,
    "temperature": 0.8,
    "max_tokens": 1024,
    "max_completion_tokens": 512
}

response_llama_4_maverick = llm_model_response(text, params)
params = {
    "model": "google/gemma-3-4b-it",
    "api_key": OPENROUTER_API_KEY,
    "base_url": OPENROUTER_DEFAULT_BASE_URL,
    "temperature": 0.8,
    "max_tokens": 1024,
    "max_completion_tokens": 512
}

text = "What is the capital of Canada? Reply only with a JSON object containing the keys 'country' and 'capital'. For example: { \"country\": \"Canada\", \"capital\": \"Ottawa\" }"
response_gemma_3_4b_it = llm_model_response(text, params)


print("Response:", response_gpt_4o_mini.content)
print("Response Metadata: ", dumps(response_gpt_4o_mini.response_metadata, indent=4))
print("Response:", response_llama_4_maverick.content)
print("Response Metadata: ", dumps(response_llama_4_maverick.response_metadata, indent=4))
print("Response:", dumps(response_gemma_3_4b_it.content, indent=4))
print("Response Metadata: ", dumps(response_gemma_3_4b_it.response_metadata, indent=4))

```
