```python
import os
from dotenv import load_dotenv

load_dotenv()

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_DEFAULT_BASE_URL = "https://openrouter.ai/api/v1"
##OPENROUTER_MODEL = "google/gemini-3.1-flash-lite-preview"
##OPENROUTER_MODEL = "ibm-granite/granite-4.1-8b"
##OPENROUTER_MODEL = "deepseek/deepseek-v4-pro"
OPENROUTER_MODEL = "openai/gpt-4o-mini"
```

```python
import langchain_openrouter
from rich.console import Console
from rich.markdown import Markdown
```

```python
## list all the parameters that can be used to create a chat model
help(langchain_openrouter.ChatOpenRouter)
```

# Create Model

```python
def llm_model(prompt_text, params=None):

    # 1. Define sensible defaults
    config = {
        "model": OPENROUTER_MODEL,
        "api_key": OPENROUTER_API_KEY,
        "base_url": OPENROUTER_DEFAULT_BASE_URL,
        "temperature": 0.5,
        "max_tokens": 256,
        "max_new_tokens": 128,
        "min_new_tokens": 10
    }
    
    # 2. Update with user params if provided
    if params:
        # Check for 'api_base' and map it to 'base_url' for consistency
        if "api_base" in params:
            params["base_url"] = params.pop("api_base")
        config.update(params)
        
    # 3. Initialize the model
    model = langchain_openrouter.ChatOpenRouter(
        model=config["model"],
        api_key=config["api_key"],
        base_url=config["base_url"],
        temperature=config["temperature"],
        max_tokens=config["max_tokens"]
    )

    response = model.invoke(prompt_text)

    return response
```

## Basic Prompt

```python
params = {
    "model": OPENROUTER_MODEL,
    "api_key": OPENROUTER_API_KEY,
    "base_url": OPENROUTER_DEFAULT_BASE_URL,
    "temperature": 0.5,
    "max_tokens": 1024,
    "max_new_tokens": 512,
    "min_new_tokens": 10
}

prompt = "The wind is "

response = llm_model(prompt, params)

Console().print(response.response_metadata)
Console().print(Markdown(f"**Prompt:** {prompt}"))
Console().print(Markdown(response.content))
```
